```python
import pandas as pd
from pathlib import Path
from sqlalchemy import create_engine
```

```python
# Use a local demo database so this works without a MySQL server.
engine = create_engine("sqlite:///:memory:")
pd.DataFrame({"name": ["Cafe One", "Food House"], "rating": [4.5, 4.2]}).to_sql("restaurants", engine, index=False)
pd.DataFrame({"name": ["Film A", "Film B"], "rating": [8.5, 7.2]}).to_sql("movies", engine, index=False)
restaurants = pd.read_sql("restaurants", engine)
restaurants.head()
```

```python
query = "SELECT name, rating FROM movies WHERE rating > 8"
print(pd.read_sql_query(query, engine))
```

```python
users = pd.read_json("https://jsonplaceholder.typicode.com/users")
print(users["username"])
```

```python
orders_file, users_file = Path("orders.csv"), Path("users.csv")
orders = pd.read_csv(orders_file) if orders_file.exists() else pd.DataFrame({"order_id": [1], "user_id": [101], "amount": [250]})
users = pd.read_csv(users_file) if users_file.exists() else pd.DataFrame({"user_id": [101], "username": ["sam"]})
print(orders.merge(users, on="user_id")[["username", "amount"]])
```

```python
today_orders = pd.DataFrame({"order_id": [1], "item": ["Book"], "price": [10]})
yesterday_orders = pd.DataFrame({"order_id": [2], "item": ["Pen"], "price": [2]})
print(pd.concat([today_orders, yesterday_orders]).reset_index(drop=True))
```
