Task 4

```python
file_path = "flipkart_product_listings_12000.xlsx"
chunk_size = 2000
start_row = 0

while True:
    chunk = pd.read_excel(
        file_path,
        skiprows=range(1, start_row + 1),
        nrows=chunk_size
    )

    if chunk.empty:
        break

    print(f"Rows in this chunk: {len(chunk)}")
    start_row += len(chunk)
    
```

Task 5

```python
data = {
    "Transaction_ID": [101, 102, 103, 104],
    "Customer": ["Rahul", "Priya", None, "Amit"],
    "Amount": [500, None, 250, 700],
    "Payment_Method": ["UPI", "Card", "Wallet", None]
}

df = pd.DataFrame(data)

# Create a semicolon-separated CSV file
df.to_csv("paytm_transactions.csv", sep=";", index=False)

print("CSV file created successfully!")
```

```python
# Read the semicolon-separated CSV file
df = pd.read_csv("paytm_transactions.csv", sep=";")

# Check missing values in each column
null_counts = df.isnull().sum()

# Print columns that contain null values
print("Columns with missing values:")

print(null_counts[null_counts > 0])
```
