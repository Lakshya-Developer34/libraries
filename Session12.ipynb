```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from pathlib import Path
```

# Task 1

```python
penguins = sns.load_dataset('penguins')
penguins = penguins.dropna()
print(penguins.shape)

fig, ax = plt.subplots(figsize=(8, 5))
sns.histplot(data=penguins, x='body_mass_g', hue='species', bins=25, kde=True, ax=ax)
ax.set_title('Body Mass by Species')
fig.savefig('penguin_hist.png')
plt.show()
```

# Task 2

```python
fig, ax = plt.subplots(figsize=(8, 5))
axes_result = sns.kdeplot(data=penguins, x='flipper_length_mm', hue='species', fill=True, ax=ax)
ax.set_title('Flipper Length (axes-level)')
fig.savefig('flipper_axes_level.png')
plt.show()
print(type(axes_result))

figure_result = sns.displot(data=penguins, x='flipper_length_mm', kind='kde', hue='species', fill=True, col='island')
figure_result.savefig('flipper_figure_level.png')
plt.show()
print(type(figure_result))
```

# Task 3

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

sns.countplot(data=penguins, x='island', hue='species', ax=axes[0])
axes[0].set_title('Penguins per Island')

sns.boxplot(data=penguins, x='species', y='body_mass_g', ax=axes[1])
axes[1].set_title('Body Mass by Species')

sns.violinplot(data=penguins, x='sex', y='bill_length_mm', ax=axes[2])
axes[2].set_title('Bill Length by Sex')

plt.tight_layout()
fig.savefig('penguin_categorical.png')
plt.show()
```

# Task 4

```python
def make_panel(style, filename):
    sns.set_theme(style=style)
    fig, axes = plt.subplots(1, 3, figsize=(15, 4))
    sns.countplot(data=penguins, x='island', hue='species', ax=axes[0])
    axes[0].set_title('Penguins per Island')
    sns.boxplot(data=penguins, x='species', y='body_mass_g', ax=axes[1])
    axes[1].set_title('Body Mass by Species')
    sns.violinplot(data=penguins, x='sex', y='bill_length_mm', ax=axes[2])
    axes[2].set_title('Bill Length by Sex')
    plt.tight_layout()
    fig.savefig(filename)
    plt.show()

make_panel('darkgrid', 'penguin_darkgrid.png')
make_panel('whitegrid', 'penguin_whitegrid.png')

for f in ['penguin_darkgrid.png', 'penguin_whitegrid.png']:
    print(f, Path(f).exists())
```
