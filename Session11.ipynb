```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# Task 1

```python
autos = pd.read_csv('auto_trips.csv', parse_dates=['trip_date'])
daily_fare = autos.set_index('trip_date')['fare_inr'].resample('D').sum()
area_counts = autos['area'].value_counts()

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

axes[0, 0].plot(daily_fare.index, daily_fare.values)
axes[0, 0].set_title('Daily Total Fare')

axes[0, 1].scatter(autos['distance_km'], autos['fare_inr'], alpha=0.6)
axes[0, 1].set_title('Distance vs Fare')

axes[1, 0].bar(area_counts.index, area_counts.values)
axes[1, 0].set_title('Trips per Area')

axes[1, 1].hist(autos['fare_inr'], bins=15, edgecolor='black')
axes[1, 1].set_title('Fare Histogram')

plt.tight_layout()
plt.savefig('trip_grid.png')
plt.show()
```

# Task 2

```python
daily_count = autos.set_index('trip_date')['fare_inr'].resample('D').count()

fig, ax1 = plt.subplots(figsize=(10, 5))
ax1.plot(daily_fare.index, daily_fare.values, color='tab:blue', linewidth=2, marker='o', label='Total fare')
ax1.set_xlabel('Date')
ax1.set_ylabel('Total Fare (Rs)', color='tab:blue')

ax2 = ax1.twinx()
ax2.plot(daily_count.index, daily_count.values, color='tab:red', linestyle='--', marker='s', label='Trip count')
ax2.set_ylabel('Trip Count', color='tab:red')

handles1, labels1 = ax1.get_legend_handles_labels()
handles2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(handles1 + handles2, labels1 + labels2, loc='upper left')

plt.title('Daily Fare vs Trip Count')
plt.savefig('fare_vs_trips_dual.png')
plt.show()
```

# Task 3

```python
areas = ['Navrangpura', 'Maninagar', 'Satellite', 'Vastrapur', 'Paldi']
fare_lists = [autos.loc[autos['area'] == a, 'fare_inr'] for a in areas]

fig, ax = plt.subplots(figsize=(8, 5))
ax.boxplot(fare_lists)
ax.set_xticklabels(areas)
ax.set_title('Fare by Area')
ax.set_xlabel('Area')
ax.set_ylabel('Fare (Rs)')
plt.savefig('fare_by_area_box.png')
plt.show()
```

# Task 4

```python
autos = pd.read_csv('auto_trips.csv', parse_dates=['trip_date'])
daily_fare = autos.set_index('trip_date')['fare_inr'].resample('D').sum()

peak_date = daily_fare.idxmax()
peak_fare = daily_fare.max()
average_fare = daily_fare.mean()

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(daily_fare.index, daily_fare.values)
ax.annotate('Highest fare day', xy=(peak_date, peak_fare), xytext=(peak_date - pd.Timedelta(days=6), peak_fare * 0.9), arrowprops=dict(arrowstyle='->'))
ax.text(peak_date, peak_fare, 'Rs ' + str(int(peak_fare)))
ax.axhline(average_fare, color='red', linestyle='--', label='Average daily total')
ax.set_title('Daily Total Fare with Peak')
ax.set_xlabel('Date')
ax.set_ylabel('Total Fare (Rs)')
ax.legend()

annotated_date = peak_date
assert annotated_date == daily_fare.idxmax()

plt.savefig('fare_peak_annotated.png')
plt.show()
```
